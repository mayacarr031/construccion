## 3. Carga de Datos con Pandas
Cargamos el dataset utilizando la función modular `load_data()` asignando encabezados apropiados.

```python
# Cargar dataset
df_raw = load_data('data/twitter_validation.csv')
df_raw.head()
```

## 4. Limpieza de Datos (Eliminación de Nulos en Columna de Texto)
Filtramos y eliminamos cualquier valor vacíos/nulo en la columna `Tweet` mediante `clean_data()`.

```python
# Eliminar filas con nulos en la columna 'Tweet'
df_clean = clean_data(df_raw, text_column='Tweet')
print(f"Total de registros limpios: {len(df_clean)}")
```

## 5. Resumen Estadístico Básico del Dataset
Generación e impresión del informe estadístico descriptivo utilizando `generate_summary()` y `print_eda_report()`.

```python
# Generar resumen estadístico
summary = generate_summary(df_clean, text_column='Tweet')
print_eda_report(summary)
```
